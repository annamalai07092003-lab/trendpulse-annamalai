```python
import time
from datetime import datetime
import requests

# --------------------------------------------------------------------------
# Configuration and Constants
# --------------------------------------------------------------------------
# API Endpoint URL patterns for HackerNews
TOP_STORIES_URL = "https://hacker-news.firebaseio.com/v0/topstories.json"
ITEM_DETAIL_URL = "https://hacker-news.firebaseio.com/v0/item/{id}.json"

# Mandatory request header to comply with the assignment requirements
HEADERS = {"User-Agent": "TrendPulse/1.0"}

# Keywords map for classifying HackerNews story titles (Case-Insensitive)
KEYWORDS_MAP = {
    "technology": ["AI", "software", "tech", "code", "computer", "data", "cloud", "API", "GPU", "LLM"],
    "worldnews": ["war", "government", "country", "president", "election", "climate", "attack", "global"],
    "sports": ["NFL", "NBA", "FIFA", "sport", "game", "team", "player", "league", "championship"],
    "science": ["research", "study", "space", "physics", "biology", "discovery", "NASA", "genome"],
    "entertainment": ["movie", "film", "music", "Netflix", "game", "book", "show", "award", "streaming"]
}

# --------------------------------------------------------------------------
# Helper Functions
# --------------------------------------------------------------------------
def determine_category(title):
    """
    Checks if any defined keywords are contained in the story title.
    Returns the matching category string if found, otherwise returns None.
    """
    if not title:
        return None

    title_lower = title.lower()

    # Iterate through each category and its associated list of keywords
    for category, keywords in KEYWORDS_MAP.items():
        for keyword in keywords:
            # Case-insensitive substring match
            if keyword.lower() in title_lower:
                return category

    return None

def fetch_top_story_ids():
    """
    Fetches the list of top story IDs from HackerNews.
    Truncates and returns the first 500 IDs. Returns an empty list if it fails.
    """
    try:
        print("Fetching top stories list from HackerNews...")
        response = requests.get(TOP_STORIES_URL, headers=HEADERS, timeout=10)

        if response.status_code == 200:
            story_ids = response.json()
            # Enforce the requirement to only look at the first 500 items
            return story_ids[:500]
        else:
            print(f"Error: Unable to fetch top stories. Status code: {response.status_code}")
            return []

    except requests.RequestException as e:
        print(f"Network error while fetching top story IDs: {e}")
        return []

# --------------------------------------------------------------------------
# Main Execution Pipeline
# --------------------------------------------------------------------------
def main():
    # Step 1: Get the list of the first 500 top story IDs
    top_ids = fetch_top_story_ids()
    if not top_ids:
        print("Pipeline aborted: Failed to retrieve top story IDs.")
        return

    # Containers to manage categorized items and enforce quota limits
    # Limits collection to a maximum of 25 stories per category
    collected_by_category = {cat: [] for cat in KEYWORDS_MAP.keys()}
    all_extracted_stories = []

    print(f"Processing up to 500 items to fill category quotas...")

    # We iterate category by category to correctly place the 2-second sleep delay
    # between separate category loops.
    for index, (category_name, keywords) in enumerate(KEYWORDS_MAP.items()):

        # Enforce requirement: Wait 2 seconds between each category loop
        # (Skip the sleep before processing the very first category)
        if index > 0:
            print(f"Pausing for 2 seconds before shifting to category: {category_name}...")
            time.sleep(2)

        print(f"Scanning for category: '{category_name}'...")

        for story_id in top_ids:
            # Stop examining items if the current category's quota is already satisfied
            if len(collected_by_category[category_name]) >= 25:
                break

            try:
                # Fetch details for the individual story ID
                url = ITEM_DETAIL_URL.format(id=story_id)
                response = requests.get(url, headers=HEADERS, timeout=5)

                if response.status_code != 200:
                    print(f"Skipping story {story_id}: HTTP Status {response.status_code}")
                    continue

                story_data = response.json()
                if not story_data:
                    continue

                # Extract the title safely to check its classification
                title = story_data.get("title", "")
                assigned_cat = determine_category(title)

                # Check if the text matches the current loop's category filter
                if assigned_cat == category_name:
                    # Parse and format fields explicitly according to requirements
                    extracted_story = {
                        "post_id": story_data.get("id"),
                        "title": title,
                        "category": assigned_cat,
                        "score": story_data.get("score", 0),
                        "num_comments": story_data.get("descendants", 0), # mapping 'descendants' to 'num_comments'
                        "author": story_data.get("by", "unknown"),
                        "collected_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
                    }

                    # Track item in our collections
                    collected_by_category[category_name].append(extracted_story)
                    all_extracted_stories.append(extracted_story)

            except requests.RequestException as e:
                # Requirement: Print message and move on if a request fails — do not crash the script
                print(f"Warning: Failed to fetch item {story_id}. Error: {e}")
                continue

    # Step 3: Ensure data/ directory exists and write out the JSON file
    os.makedirs("data", exist_ok=True)

    # Dynamic filename structure based on the exact running date (trends_YYYYMMDD.json)
    current_date_str = datetime.now().strftime("%Y%m%d")
    output_filepath = f"data/trends_{current_date_str}.json"

    try:
        import json
        with open(output_filepath, "w", encoding="utf-8") as f:
            json.dump(all_extracted_stories, f, indent=4, ensure_ascii=False)

        # Target console log validation output
        print(f"Collected {len(all_extracted_stories)} stories. Saved to {output_filepath}")

        # Display granular categorical progress for verification
        print("\nBreakdown by category:")
        for cat, items in collected_by_category.items():
            print(f" - {cat}: {len(items)}/25 collected")

    except IOError as e:
        print(f"Critical error writing the JSON data file: {e}")

if __name__ == "__main__":
    main()
```

```python
# @title Default title text
import os
import csv
import json
import glob
from datetime import datetime

# --------------------------------------------------------------------------
# Configuration and Constants
# --------------------------------------------------------------------------
DATA_DIR = "data"
# Expected columns in the target clean CSV file
CSV_COLUMNS = ["post_id", "title", "category", "score", "num_comments", "author", "collected_at"]

def find_latest_json_file(directory):
    """
    Scans the data directory for files matching the pattern trends_*.json
    and returns the path of the most recent one.
    """
    search_pattern = os.path.join(directory, "trends_*.json")
    json_files = glob.glob(search_pattern)

    if not json_files:
        return None

    # Sort files naturally by name (since YYYYMMDD sorts chronologically)
    json_files.sort()
    return json_files[-1]

def clean_and_convert():
    """
    Loads the latest raw JSON file, runs data cleaning/deduplication,
    and writes the output to a cleaned CSV.
    """
    # 1. Locate the input JSON file
    input_file = find_latest_json_file(DATA_DIR)
    if not input_file:
        print(f"Error: No raw JSON trend files found in the '{DATA_DIR}/' folder.")
        print("Please ensure Task 1 has run successfully.")
        return

    print(f"Found latest raw data file: {input_file}")

    try:
        with open(input_file, "r", encoding="utf-8") as f:
            raw_stories = json.load(f)
    except (json.JSONDecodeError, IOError) as e:
        print(f"Critical Error reading/parsing {input_file}: {e}")
        return

    print(f"Loaded {len(raw_stories)} raw items. Beginning cleaning pipeline...")

    # Tracking sets for data normalization and quality checks
    seen_post_ids = set()
    cleaned_records = []

    # Counter variables to report data health statistics at the end
    duplicate_count = 0
    missing_fields_fixed = 0

    # 2. Iterate and Clean Records
    for item in raw_stories:
        # --- Deduplication Check ---
        post_id = item.get("post_id")

        # Drop item if it doesn't contain a valid ID or if it is a duplicate
        if post_id is None:
            continue

        if post_id in seen_post_ids:
            duplicate_count += 1
            continue
        seen_post_ids.add(post_id)

        # --- Missing Values & Data Type Normalization ---
        # Safeguard numeric properties against missing values or string corruptions
        try:
            score = int(item.get("score", 0))
        except (ValueError, TypeError):
            score = 0
            missing_fields_fixed += 1

        try:
            num_comments = int(item.get("num_comments", 0))
        except (ValueError, TypeError):
            num_comments = 0
            missing_fields_fixed += 1

        # Strip whitespace out of strings and provide logical fallbacks
        title = str(item.get("title", "")).strip()
        category = str(item.get("category", "uncategorized")).strip()
        author = str(item.get("author", "unknown")).strip()
        collected_at = str(item.get("collected_at", "")).strip()

        if not author:
            author = "unknown"

        if not collected_at:
            collected_at = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        # Package the validated data
        clean_record = {
            "post_id": post_id,
            "title": title,
            "category": category,
            "score": score,
            "num_comments": num_comments,
            "author": author,
            "collected_at": collected_at
        }
        cleaned_records.append(clean_record)

    # 3. Save Structured Dataset to a Clean CSV
    current_date_str = datetime.now().strftime("%Y%m%d")
    output_filename = f"cleaned_trends_{current_date_str}.csv"
    output_filepath = os.path.join(DATA_DIR, output_filename)

    try:
        with open(output_filepath, "w", newline="", encoding="utf-8") as csv_file:
            writer = csv.DictWriter(csv_file, fieldnames=CSV_COLUMNS)

            # Write structured headers
            writer.writeheader()
            # Write rows sequentially
            writer.writerows(cleaned_records)

        # Pipeline Terminal Reports
        print("\n--- Cleaning Report Summary ---")
        print(f"Total Rows Evaluated:   {len(raw_stories)}")
        print(f"Duplicates Removed:     {duplicate_count}")
        print(f"Corrupted Fields Fixed: {missing_fields_fixed}")
        print(f"Clean Rows Exported:    {len(cleaned_records)}")
        print(f"Successfully saved to:  {output_filepath}")

    except IOError as e:
        print(f"Critical error writing out the CSV file: {e}")

if __name__ == "__main__":
    clean_and_convert()
```
